Q8.

Count nested list [1,2] by count method

data = [[1, 2],[3, 4],[1, 2],[2,4],[1, 2],[1,3,6]]


```python
data = [[1, 2],[3, 4],[1, 2],[2,4],[1, 2],[1,3,6]]
count = 0 
for i in data:
    if i == [1,2]:
        count += 1 
print(count)
```

Q10.

Print all two results

Q11.
 You have a product code that follows the format "SKU-12345-XYZ".

Find the first position of the dash (-).

Count the number of dashes (-) in the product code.

Remove the "SKU-" prefix and "-XYZ" suffix from the product code, Extract the numeric part 12345

product_code = "SKU-12345-XYZ"

Hint: Use String Methods

```python
prodCode = "SKU-12345-XYZ"
dashPos = prodCode.find("-")
dashCount = prodCode.count("-")
prodNumeric = prodCode.split("-")[1]
print("position of the dash (-) is",dashPos)
print("Count of dashes (-) in the product code is",dashCount)
print("numeric part of prodCode is",prodNumeric)
```

Q14.
text = "    Data Analysis with Python is very powerful    "

Perform the following tasks:

Remove extra spaces using strip().
Convert the string into title case using title().
Replace "Powerful" with "Important".
Find the position of the word "Python".

```python
text = " Data Analysis with Python is very powerful "
updatedText = text.title().replace("Powerful","Important").strip()
print(updatedText)
print("index of python in updated string is ",updatedText.find("Python"))
```
